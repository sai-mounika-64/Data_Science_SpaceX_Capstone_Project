```python
# Cell 1: Install and Import Libraries
!pip install beautifulsoup4 requests pandas lxml html5lib

import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

# Cell 2: Define Target URL
# Wikipedia page with Falcon 9 launch records
url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
print(f"Target URL: {url}")

# Cell 3: Fetch Web Page
# Send GET request
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"}

response = requests.get(url, headers=headers)
print(f"Status Code: {response.status_code}")
print(f"Page size: {len(response.content)} bytes")

if response.status_code == 200:
    print("✓ Page fetched successfully!")
else:
    print("✗ Failed to fetch page")

if response.status_code == 200:
    print("✓ Page fetched successfully!")
else:
    print("✗ Failed to fetch page")

# Cell 4: Parse HTML Content
# Create BeautifulSoup object
soup = BeautifulSoup(response.content, 'html.parser')

# Get page title
page_title = soup.find('h1', class_='firstHeading').text
print(f"Page Title: {page_title}")

# Cell 5: Find All Tables
# Find all tables with class 'wikitable'
tables = soup.find_all('table', class_='wikitable')
print(f"Number of tables found: {len(tables)}")

# Display table headers
for i, table in enumerate(tables[:3]):  # Show first 3 tables
    print(f"\nTable {i+1} headers:")
    headers = table.find_all('th')
    print([header.text.strip() for header in headers[:5]])  # First 5 headers

# Cell 6: Extract Launch Data from Tables
# Function to extract table data
def extract_table_data(table):
    """Extract data from HTML table"""
    data = []
    
    # Get headers
    headers = []
    header_row = table.find('tr')
    for th in header_row.find_all('th'):
        headers.append(th.text.strip())
    
    # Get rows
    rows = table.find_all('tr')[1:]  # Skip header row
    
    for row in rows:
        cols = row.find_all(['td', 'th'])
        cols_data = [col.text.strip() for col in cols]
        if cols_data:  # Only add non-empty rows
            data.append(cols_data)
    
    return headers, data

# Cell 7: Process First Launch Table
# Extract data from first table (usually 2010-2013 launches)
if len(tables) > 0:
    headers_1, data_1 = extract_table_data(tables[0])
    df_1 = pd.DataFrame(data_1, columns=headers_1 if len(headers_1) == len(data_1[0]) else None)
    print(f"Table 1 shape: {df_1.shape}")
    display(df_1.head())

# Cell 8: Process All Launch Tables
# Combine all tables
all_dataframes = []

for i, table in enumerate(tables):
    try:
        headers, data = extract_table_data(table)
        if data:
            df_temp = pd.DataFrame(data)
            # Use headers if length matches
            if len(headers) == df_temp.shape[1]:
                df_temp.columns = headers
            all_dataframes.append(df_temp)
            print(f"Table {i+1}: {df_temp.shape}")
    except Exception as e:
        print(f"Error processing table {i+1}: {e}")

# Cell 9: Combine All Tables
# Concatenate all dataframes
if all_dataframes:
    combined_df = pd.concat(all_dataframes, ignore_index=True)
    print(f"\nCombined dataset shape: {combined_df.shape}")
    print(f"Columns: {combined_df.columns.tolist()}")
    display(combined_df.head())

# Cell 10: Clean Column Names
# Standardize column names
def clean_column_name(col):
    """Clean and standardize column names"""
    # Remove special characters and extra spaces
    col = re.sub(r'\[.*?\]', '', col)  # Remove references
    col = col.strip()
    col = re.sub(r'\s+', ' ', col)  # Replace multiple spaces
    return col

combined_df.columns = [clean_column_name(col) for col in combined_df.columns]
print("Cleaned columns:")
print(combined_df.columns.tolist())

# Cell 11: Extract Key Columns
# Identify and select key columns
key_columns_mapping = {
    'Flight No.': 'FlightNumber',
    'Date and time (UTC)': 'Date',
    'Launch site': 'LaunchSite',
    'Payload': 'Payload',
    'Payload mass': 'PayloadMass',
    'Orbit': 'Orbit',
    'Customer': 'Customer',
    'Launch outcome': 'LaunchOutcome',
    'Booster landing': 'BoosterLanding'
}

# Find matching columns
selected_cols = []
for old_name, new_name in key_columns_mapping.items():
    for col in combined_df.columns:
        if old_name.lower() in col.lower():
            selected_cols.append(col)
            break

print(f"Selected columns: {selected_cols}")

# Cell 12: Create Clean Dataset
# Select and rename columns
if selected_cols:
    clean_df = combined_df[selected_cols].copy()
    # Rename to standard names
    clean_df.columns = list(key_columns_mapping.values())[:len(selected_cols)]
    display(clean_df.head())

# Cell 13: Parse Date Column
# Convert date strings to datetime
if 'Date' in clean_df.columns:
    # Extract date part (remove time if present)
    clean_df['Date'] = clean_df['Date'].str.split(',').str[0]
    clean_df['Date'] = pd.to_datetime(clean_df['Date'], errors='coerce')
    clean_df['Year'] = clean_df['Date'].dt.year
    clean_df['Month'] = clean_df['Date'].dt.month
    
    print("Date parsing completed:")
    print(clean_df[['Date', 'Year', 'Month']].head())

# Cell 14: Parse Payload Mass
# Extract numeric payload mass
if 'PayloadMass' in clean_df.columns:
    # Extract numbers from payload mass string
    clean_df['PayloadMass_kg'] = clean_df['PayloadMass'].str.extract(r'(\d+(?:,\d+)?(?:\.\d+)?)')[0]
    clean_df['PayloadMass_kg'] = clean_df['PayloadMass_kg'].str.replace(',', '').astype(float)
    
    print("Payload mass parsing:")
    print(clean_df[['PayloadMass', 'PayloadMass_kg']].head())

# Cell 15: Clean Launch Site Names
# Standardize launch site names
if 'LaunchSite' in clean_df.columns:
    # Remove extra information in parentheses
    clean_df['LaunchSite_clean'] = clean_df['LaunchSite'].str.split('(').str[0].str.strip()
    
    print("Launch sites:")
    print(clean_df['LaunchSite_clean'].value_counts())

# Cell 16: Parse Landing Outcomes
# Extract landing outcome status
if 'BoosterLanding' in clean_df.columns:
    # Check for success keywords
    success_keywords = ['success', 'successful', 'landed']
    failure_keywords = ['failure', 'failed', 'crashed', 'lost', 'no attempt']
    
    def categorize_landing(landing_text):
        if pd.isna(landing_text):
            return 'Unknown'
        landing_text = str(landing_text).lower()
        if any(keyword in landing_text for keyword in success_keywords):
            return 'Success'
        elif any(keyword in landing_text for keyword in failure_keywords):
            return 'Failure'
        else:
            return 'Other'
    
    clean_df['LandingCategory'] = clean_df['BoosterLanding'].apply(categorize_landing)
    
    print("Landing outcomes:")
    print(clean_df['LandingCategory'].value_counts())

# Cell 17: Remove Duplicates and Invalid Rows
# Remove rows with missing critical data
print(f"Rows before cleaning: {len(clean_df)}")

# Remove duplicates
clean_df = clean_df.drop_duplicates()

# Remove rows with no date
clean_df = clean_df.dropna(subset=['Date'])

print(f"Rows after cleaning: {len(clean_df)}")

# Cell 18: Create Final Dataset
# Select final columns
final_columns = [
    'FlightNumber', 'Date', 'Year', 'Month',
    'LaunchSite_clean', 'Payload', 'PayloadMass_kg',
    'Orbit', 'Customer', 'LaunchOutcome', 'LandingCategory'
]

# Keep only columns that exist
final_columns = [col for col in final_columns if col in clean_df.columns]

scraped_df = clean_df[final_columns].copy()
scraped_df.columns = [col.replace('_clean', '').replace('_kg', '') for col in final_columns]

print(f"Final scraped dataset shape: {scraped_df.shape}")
display(scraped_df.head(10))

# Cell 19: Data Quality Check
print("\n" + "=" * 60)
print("DATA QUALITY SUMMARY")
print("=" * 60)
print("\nMissing values:")
print(scraped_df.isnull().sum())
print("\nData types:")
print(scraped_df.dtypes)
print(f"\nDate range: {scraped_df['Date'].min()} to {scraped_df['Date'].max()}")

# Cell 20: Save Scraped Data
# Save to CSV
scraped_df.to_csv('data/spacex_scraped_data.csv', index=False)
print("\n✓ Scraped data saved to 'data/spacex_scraped_data.csv'")

# Cell 21: Visualization
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Launches per year
scraped_df['Year'].value_counts().sort_index().plot(kind='bar', ax=axes[0, 0], color='teal')
axes[0, 0].set_title('Wikipedia: Launches per Year', fontweight='bold')
axes[0, 0].set_xlabel('Year')
axes[0, 0].set_ylabel('Count')

# Launch sites
if 'LaunchSite' in scraped_df.columns:
    scraped_df['LaunchSite'].value_counts().plot(kind='barh', ax=axes[0, 1], color='coral')
    axes[0, 1].set_title('Launches per Site', fontweight='bold')
    axes[0, 1].set_xlabel('Count')

# Landing outcomes
if 'LandingCategory' in scraped_df.columns:
    scraped_df['LandingCategory'].value_counts().plot(kind='pie', ax=axes[1, 0], autopct='%1.1f%%')
    axes[1, 0].set_title('Landing Outcomes', fontweight='bold')
    axes[1, 0].set_ylabel('')

# Payload mass distribution
if 'PayloadMass' in scraped_df.columns:
    scraped_df['PayloadMass'].hist(bins=20, ax=axes[1, 1], color='purple', alpha=0.7)
    axes[1, 1].set_title('Payload Mass Distribution', fontweight='bold')
    axes[1, 1].set_xlabel('Payload Mass (kg)')
    axes[1, 1].set_ylabel('Frequency')

plt.tight_layout()
plt.savefig('web_scraping_summary.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✓ All web scraping tasks completed!")
```
