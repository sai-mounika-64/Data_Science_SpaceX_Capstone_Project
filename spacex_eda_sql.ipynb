```python
# Cell 1: Install and Import Libraries
!pip install sqlalchemy

import pandas as pd
import numpy as np
import sqlite3
from sqlalchemy import create_engine
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")

# Cell 2: Load Clean Data
df = pd.read_csv('data/spacex_cleaned.csv')
df['Date'] = pd.to_datetime(df['Date'])

print(f"Dataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
display(df.head())

# Cell 3: Create SQLite Database
conn = sqlite3.connect('spacex.db')
print("Database connection established!")

df.to_sql('SPACEXTBL', conn, if_exists='replace', index=False)
print("Data loaded into SPACEXTBL table")

cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print(f"Tables in database: {cursor.fetchall()}")

# Cell 4: View Table Schema
query = """
PRAGMA table_info(SPACEXTBL);
"""
schema = pd.read_sql_query(query, conn)
print("Table Schema:")
display(schema)

# Cell 5: Basic Query - View All Data
query = """
SELECT * FROM SPACEXTBL LIMIT 10;
"""
result = pd.read_sql_query(query, conn)
print("First 10 rows from SPACEXTBL:")
display(result)

# Cell 6: TASK 1 - Display the names of the unique launch sites
print("=" * 70)
print("TASK 1: All Launch Site Names")
print("=" * 70)

query = """
SELECT DISTINCT LaunchSite 
FROM SPACEXTBL
ORDER BY LaunchSite;
"""

result = pd.read_sql_query(query, conn)
print("\nUnique Launch Sites:")
display(result)
print(f"\nTotal unique launch sites: {len(result)}")

plt.figure(figsize=(10, 6))
site_counts = df['LaunchSite'].value_counts()
plt.bar(site_counts.index, site_counts.values, color='steelblue', edgecolor='black')
plt.title('Launch Count by Site', fontsize=14, fontweight='bold')
plt.xlabel('Launch Site')
plt.ylabel('Number of Launches')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('sql_task1_launch_sites.png', dpi=300, bbox_inches='tight')
plt.show()

# Cell 7: TASK 2 - Display 5 records where launch sites begin with 'CCS'
print("=" * 70)
print("TASK 2: Launch Sites Beginning with 'CCS'")
print("=" * 70)

query = """
SELECT * 
FROM SPACEXTBL 
WHERE LaunchSite LIKE 'CCS%'
LIMIT 5;
"""

result = pd.read_sql_query(query, conn)
print("\n5 records where launch site begins with 'CCS':")
display(result)

query_count = """
SELECT COUNT(*) as Total_CCA_Launches
FROM SPACEXTBL 
WHERE LaunchSite LIKE 'CCA%';
"""
count_result = pd.read_sql_query(query_count, conn)
print(f"\nTotal launches from CCA sites: {count_result['Total_CCA_Launches'].iloc[0]}")

# Cell 8: TASK 3 - Display total payload mass carried by NASA (CRS)
print("=" * 70)
print("TASK 3: Total Payload Mass Carried by NASA (CRS)")
print("=" * 70)

# Note: Since we don't have Mission column, we'll use FlightNumber or check MissionOutcome
query = """
SELECT SUM(PayloadMass) as Total_Payload_Mass
FROM SPACEXTBL 
WHERE MissionOutcome LIKE '%CRS';
"""

result = pd.read_sql_query(query, conn)
print("\nTotal payload mass carried by NASA (CRS) missions:")
display(result)

if result['Total_Payload_Mass'].iloc[0] is not None:
    print(f"\nTotal Payload Mass: {result['Total_Payload_Mass'].iloc[0]:,.2f} kg")
else:
    print("\nNo NASA CRS missions found in dataset")

query_alt = """
SELECT 
    COUNT(*) as Mission_Count,
    SUM(PayloadMass) as Total_Payload_Mass,
    AVG(PayloadMass) as Average_Payload_Mass
FROM SPACEXTBL 
WHERE MissionOutcome LIKE '%CRS';
"""

result_alt = pd.read_sql_query(query_alt, conn)
print("\nDetailed NASA CRS Statistics:")
display(result_alt)

# Cell 9: TASK 4 - Display average payload mass carried by booster version F9 v1.1
print("=" * 70)
print("TASK 4: Average Payload Mass by Booster Version F9 v1.1")
print("=" * 70)

# Note: We don't have a Rocket column, so we'll calculate overall average
query = """
SELECT 
    AVG(PayloadMass) as Average_Payload_Mass,
    COUNT(*) as Mission_Count,
    MIN(PayloadMass) as Min_Payload,
    MAX(PayloadMass) as Max_Payload
FROM SPACEXTBL;
"""

result = pd.read_sql_query(query, conn)
print("\nOverall payload mass statistics:")
display(result)

if result['Average_Payload_Mass'].iloc[0] is not None:
    print(f"\nAverage Payload Mass: {result['Average_Payload_Mass'].iloc[0]:,.2f} kg")

# Cell 10: TASK 5 - Display date of first successful ground landing
print("=" * 70)
print("TASK 5: First Successful Ground Landing Date")
print("=" * 70)

query = """
SELECT 
    MIN(Date) as First_Success_Date,
    FlightNumber,
    LaunchSite,
    LandingOutcome
FROM SPACEXTBL 
WHERE LandingOutcome LIKE '%success' 
  AND (LandingOutcome LIKE '%ground' OR LandingOutcome LIKE '%rtls')
GROUP BY FlightNumber, LaunchSite, LandingOutcome
ORDER BY Date
LIMIT 1;
"""

result = pd.read_sql_query(query, conn)
print("\nFirst successful ground pad landing:")
display(result)

query_all = """
SELECT 
    Date,
    FlightNumber,
    LaunchSite,
    LandingOutcome
FROM SPACEXTBL 
WHERE LandingOutcome LIKE '%success' 
  AND (LandingOutcome LIKE '%ground' OR LandingOutcome LIKE '%rtls')
ORDER BY Date
LIMIT 5;
"""

result_all = pd.read_sql_query(query_all, conn)
print("\nFirst 5 successful ground pad landings:")
display(result_all)

# Cell 11: TASK 6 - Successful drone ship landings with payload 4000-6000 kg
print("=" * 70)
print("TASK 6: Successful Drone Ship Landings with Payload 4000-6000 kg")
print("=" * 70)

query = """
SELECT 
    FlightNumber,
    PayloadMass,
    LandingOutcome,
    Date,
    LaunchSite
FROM SPACEXTBL 
WHERE LandingOutcome LIKE '%success' 
  AND (LandingOutcome LIKE '%drone' OR LandingOutcome LIKE '%asds' OR LandingOutcome LIKE '%ocean')
  AND PayloadMass > 4000 
  AND PayloadMass < 6000
ORDER BY Date;
"""

result = pd.read_sql_query(query, conn)
print("\nSuccessful drone ship landings with payload between 4000-6000 kg:")
display(result)
print(f"\nTotal missions matching criteria: {len(result)}")

if len(result) > 0:
    plt.figure(figsize=(12, 6))
    plt.scatter(result['Date'], result['PayloadMass'], 
                s=100, alpha=0.6, c='green', edgecolors='black')
    plt.axhline(y=4000, color='r', linestyle='--', label='Min Payload (4000 kg)')
    plt.axhline(y=6000, color='r', linestyle='--', label='Max Payload (6000 kg)')
    plt.title('Successful Drone Ship Landings (Payload 4000-6000 kg)', 
              fontsize=14, fontweight='bold')
    plt.xlabel('Date')
    plt.ylabel('Payload Mass (kg)')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('sql_task6_drone_ship_payload.png', dpi=300, bbox_inches='tight')
    plt.show()

# Cell 12: TASK 7 - Total successful and failure mission outcomes
print("=" * 70)
print("TASK 7: Total Successful and Failure Mission Outcomes")
print("=" * 70)

query = """
SELECT 
    Class,
    COUNT(*) as Count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM SPACEXTBL), 2) as Percentage
FROM SPACEXTBL 
GROUP BY Class
ORDER BY Class;
"""

result = pd.read_sql_query(query, conn)
print("\nMission outcomes summary:")
display(result)

result['Outcome'] = result['Class'].map({0: 'Failure', 1: 'Success'})
print("\nLabeled outcomes:")
display(result[['Outcome', 'Count', 'Percentage']])

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].bar(result['Outcome'], result['Count'], 
            color=['#e74c3c', '#2ecc71'], edgecolor='black', linewidth=2)
axes[0].set_title('Mission Outcomes Count', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Number of Missions')
axes[0].grid(axis='y', alpha=0.3)
for i, (outcome, count) in enumerate(zip(result['Outcome'], result['Count'])):
    axes[0].text(i, count + 1, str(count), ha='center', fontweight='bold', fontsize=12)

axes[1].pie(result['Count'], labels=result['Outcome'], autopct='%1.1f%%',
            colors=['#e74c3c', '#2ecc71'], startangle=90, 
            textprops={'fontsize': 12, 'fontweight': 'bold'})
axes[1].set_title('Mission Outcomes Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('sql_task7_mission_outcomes.png', dpi=300, bbox_inches='tight')
plt.show()

# Cell 13: TASK 8 - Boosters with maximum payload mass
print("=" * 70)
print("TASK 8: Missions with Maximum Payload")
print("=" * 70)

query_max = """
SELECT MAX(PayloadMass) as Max_Payload
FROM SPACEXTBL;
"""

max_payload = pd.read_sql_query(query_max, conn)
print(f"Maximum payload mass: {max_payload['Max_Payload'].iloc[0]:,.2f} kg")

query = """
SELECT 
    FlightNumber,
    PayloadMass,
    Date,
    LaunchSite,
    Orbit_clean as Orbit,
    Class
FROM SPACEXTBL 
WHERE PayloadMass = (SELECT MAX(PayloadMass) FROM SPACEXTBL);
"""

result = pd.read_sql_query(query, conn)
print("\nMissions with maximum payload mass:")
display(result)

query_top10 = """
SELECT 
    FlightNumber,
    PayloadMass,
    Date,
    LaunchSite,
    Class
FROM SPACEXTBL 
ORDER BY PayloadMass DESC
LIMIT 10;
"""

result_top10 = pd.read_sql_query(query_top10, conn)
print("\nTop 10 heaviest payloads:")
display(result_top10)

plt.figure(figsize=(12, 8))
plt.barh(range(len(result_top10)), result_top10['PayloadMass'], 
         color='purple', edgecolor='black')
plt.yticks(range(len(result_top10)), result_top10['FlightNumber'])
plt.xlabel('Payload Mass (kg)', fontsize=12)
plt.title('Top 10 Heaviest Payloads', fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
for i, mass in enumerate(result_top10['PayloadMass']):
    plt.text(mass + 200, i, f'{mass:,.0f} kg', va='center', fontweight='bold')
plt.tight_layout()
plt.savefig('sql_task8_max_payload.png', dpi=300, bbox_inches='tight')
plt.show()

# Cell 14: TASK 9 - 2015 failed drone ship landings
print("=" * 70)
print("TASK 9: 2015 Failed Drone Ship Landings")
print("=" * 70)

query = """
SELECT 
    Date,
    FlightNumber,
    LaunchSite,
    LandingOutcome,
    PayloadMass
FROM SPACEXTBL 
WHERE strftime('%Y', Date) = '2015'
  AND Class = 0
  AND (LandingOutcome LIKE '%drone%' OR LandingOutcome LIKE '%asds%' OR LandingOutcome LIKE '%ocean%')
ORDER BY Date;
"""

result = pd.read_sql_query(query, conn)
print("\nFailed drone ship landings in 2015:")
display(result)

if len(result) > 0:
    print(f"\nTotal failed drone ship landings in 2015: {len(result)}")
else:
    print("\nNo failed drone ship landings in 2015")

query_all_2015 = """
SELECT 
    Date,
    FlightNumber,
    LaunchSite,
    Class,
    LandingOutcome
FROM SPACEXTBL 
WHERE strftime('%Y', Date) = '2015'
ORDER BY Date;
"""

result_all = pd.read_sql_query(query_all_2015, conn)
print("\nAll 2015 launches:")
display(result_all)

# Cell 15: TASK 10 - Ranked landing outcomes (2010-2017)
print("=" * 70)
print("TASK 10: Ranked Landing Outcomes (2010-06-04 to 2017-03-20)")
print("=" * 70)

query = """
SELECT 
    LandingOutcome,
    COUNT(*) as Count
FROM SPACEXTBL 
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY LandingOutcome
ORDER BY Count DESC;
"""

result = pd.read_sql_query(query, conn)
print("\nLanding outcomes ranked by count:")
display(result)

result['Percentage'] = (result['Count'] / result['Count'].sum() * 100).round(2)
print("\nWith percentages:")
display(result)

plt.figure(figsize=(12, 8))
plt.barh(range(len(result)), result['Count'], color='teal', edgecolor='black')
plt.yticks(range(len(result)), result['LandingOutcome'])
plt.xlabel('Count', fontsize=12)
plt.title('Landing Outcomes Count (2010-06-04 to 2017-03-20)', 
          fontsize=14, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

for i, (count, pct) in enumerate(zip(result['Count'], result['Percentage'])):
    plt.text(count + 0.5, i, f'{count} ({pct}%)', va='center', fontweight='bold')

plt.tight_layout()
plt.savefig('sql_task10_ranked_outcomes.png', dpi=300, bbox_inches='tight')
plt.show()

# Cell 16: ADDITIONAL - Success rate by launch site
print("=" * 70)
print("ADDITIONAL ANALYSIS: Success Rate by Launch Site")
print("=" * 70)

query = """
SELECT 
    LaunchSite,
    COUNT(*) as Total_Launches,
    SUM(Class) as Successful_Landings,
    ROUND(AVG(Class) * 100, 2) as Success_Rate
FROM SPACEXTBL 
GROUP BY LaunchSite
ORDER BY Success_Rate DESC;
"""

result = pd.read_sql_query(query, conn)
print("\nSuccess rate by launch site:")
display(result)

plt.figure(figsize=(12, 6))
plt.bar(result['LaunchSite'], result['Success_Rate'], 
        color='coral', edgecolor='black', linewidth=1.5)
plt.title('Success Rate by Launch Site', fontsize=14, fontweight='bold')
plt.xlabel('Launch Site')
plt.ylabel('Success Rate (%)')
plt.ylim([0, 110])
plt.grid(axis='y', alpha=0.3)

for i, (site, rate, total) in enumerate(zip(result['LaunchSite'], 
                                              result['Success_Rate'], 
                                              result['Total_Launches'])):
    plt.text(i, rate + 3, f'{rate}%\n(n={total})', ha='center', fontweight='bold')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('sql_additional_site_success.png', dpi=300, bbox_inches='tight')
plt.show()

# Cell 17: Close connection
conn.close()
print("\n[DONE] Database connection closed")
print("[DONE] ALL SQL ANALYSIS TASKS COMPLETED!")
```
