```python
# Cell 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")

# Cell 2: Load Data
# Load the API data
df_api = pd.read_csv('data/spacex_launch_data.csv')
print(f"API Data shape: {df_api.shape}")
print(f"Columns: {df_api.columns.tolist()}")

# Cell 3: Display First Few Rows
print("First 10 rows of data:")
display(df_api.head(10))

# Cell 4: Data Overview
print("\n" + "=" * 60)
print("DATA OVERVIEW")
print("=" * 60)
print(f"Total records: {len(df_api)}")
print(f"Total columns: {len(df_api.columns)}")
print(f"\nColumn names and types:")
print(df_api.dtypes)

# Cell 5: Check for Missing Values
print("\n" + "=" * 60)
print("MISSING VALUES ANALYSIS")
print("=" * 60)

missing_values = df_api.isnull().sum()
missing_percent = (missing_values / len(df_api)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing_values,
    'Missing_Percent': missing_percent
})
missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)

print(missing_df)

# Visualize missing values
if len(missing_df) > 0:
    plt.figure(figsize=(10, 6))
    missing_df['Missing_Percent'].plot(kind='barh', color='salmon')
    plt.xlabel('Missing Percentage (%)')
    plt.title('Missing Values by Column', fontweight='bold')
    plt.tight_layout()
    plt.show()

# Cell 6: Handle Missing Values in Payload Mass
# Check PayloadMass distribution
print("\nPayloadMass statistics:")
print(df_api['PayloadMass'].describe())

# Fill missing payload mass with median
median_payload = df_api['PayloadMass'].median()
df_api['PayloadMass'].fillna(median_payload, inplace=True)

print(f"\n✓ Filled {missing_values['PayloadMass']} missing PayloadMass values with median: {median_payload}")

# Cell 7: Handle Missing Values in Orbit
# Check unique orbits
print("\nUnique orbits:")
print(df_api['Orbit'].value_counts())

# Fill missing orbits with 'Unknown'
df_api['Orbit'].fillna('Unknown', inplace=True)
print("\n✓ Filled missing Orbit values with 'Unknown'")

# Cell 8: Handle Missing Values in Landing Outcome
# Check landing outcomes
print("\nLanding outcomes:")
print(df_api['LandingOutcome'].value_counts())

# Fill missing landing outcomes
df_api['LandingOutcome'].fillna('No attempt', inplace=True)
print("\n✓ Filled missing LandingOutcome values")

# Cell 9: Check for Duplicates
# Check for duplicate rows
duplicates = df_api.duplicated().sum()
print(f"\nDuplicate rows found: {duplicates}")

if duplicates > 0:
    df_api = df_api.drop_duplicates()
    print(f"✓ Removed {duplicates} duplicate rows")

# Cell 10: Data Type Conversions
# Convert Date to datetime
df_api['Date'] = pd.to_datetime(df_api['Date'])

# Ensure numeric columns are correct type
df_api['FlightNumber'] = df_api['FlightNumber'].astype(int)
df_api['PayloadMass'] = pd.to_numeric(df_api['PayloadMass'], errors='coerce')
df_api['Year'] = df_api['Year'].astype(int)
df_api['Month'] = df_api['Month'].astype(int)

print("\n✓ Data types converted")
print(df_api.dtypes)

# Cell 11: Create Class Column (Target Variable)
# Create binary classification target
# 1 = Successful landing, 0 = Failed landing or no attempt

def create_class(landing_outcome):
    """Create binary class for landing success"""
    if pd.isna(landing_outcome):
        return 0
    
    # Convert to string and lowercase
    landing_outcome = str(landing_outcome).lower()
    
    # Define success keywords
    success_keywords = [
        'success', 
        'true rtls', 
        'true ocean', 
        'true asds',
        'landed',
        'successful'
    ]
    
    # Check if any success keyword is in the landing outcome
    if any(keyword in landing_outcome for keyword in success_keywords):
        return 1
    else:
        return 0

# Apply the function
df_api['Class'] = df_api['LandingOutcome'].apply(create_class)

# Verify the creation
print("=" * 60)
print("CLASS COLUMN CREATED")
print("=" * 60)
print("\nClass distribution:")
print(df_api['Class'].value_counts())

if len(df_api['Class'].unique()) > 0:
    print(f"\nSuccess rate: {df_api['Class'].mean() * 100:.2f}%")
    print(f"Total Successes: {(df_api['Class'] == 1).sum()}")
    print(f"Total Failures: {(df_api['Class'] == 0).sum()}")
else:
    print("\n⚠️ WARNING: Class column has no valid values!")

# Show some examples
print("\nSample landing outcomes and their classes:")
sample_df = df_api[['LandingOutcome', 'Class']].drop_duplicates().head(10)
print(sample_df.to_string(index=False))
print("=" * 60)

# Cell 12: Handle Outliers in Payload Mass
# Visualize payload mass distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot
df_api.boxplot(column='PayloadMass', ax=axes[0])
axes[0].set_title('Payload Mass - Boxplot', fontweight='bold')
axes[0].set_ylabel('Payload Mass (kg)')

# Histogram
df_api['PayloadMass'].hist(bins=30, ax=axes[1], color='steelblue', edgecolor='black')
axes[1].set_title('Payload Mass - Distribution', fontweight='bold')
axes[1].set_xlabel('Payload Mass (kg)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

# Identify outliers using IQR method
Q1 = df_api['PayloadMass'].quantile(0.25)
Q3 = df_api['PayloadMass'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_api[(df_api['PayloadMass'] < lower_bound) | (df_api['PayloadMass'] > upper_bound)]
print(f"\nOutliers found: {len(outliers)}")
print(f"Outlier range: < {lower_bound:.2f} or > {upper_bound:.2f} kg")

# Keep outliers as they represent valid heavy payloads
print("Note: Keeping outliers as they represent legitimate heavy payloads")

# Cell 13: Clean Launch Site Names
# Standardize launch site names
print("\nOriginal launch sites:")
print(df_api['LaunchSite'].value_counts())

# Clean launch site names
df_api['LaunchSite'] = df_api['LaunchSite'].str.strip()
df_api['LaunchSite'] = df_api['LaunchSite'].str.replace('CCAFS', 'CCAFS')
df_api['LaunchSite'] = df_api['LaunchSite'].str.replace('KSC', 'KSC')
df_api['LaunchSite'] = df_api['LaunchSite'].str.replace('VAFB', 'VAFB')

print("\nCleaned launch sites:")
print(df_api['LaunchSite'].value_counts())

# Cell 14: Clean Orbit Names
# Standardize orbit names
print("\nOriginal orbits:")
print(df_api['Orbit'].value_counts())

# Group similar orbits
orbit_mapping = {
    'LEO': 'LEO',
    'GTO': 'GTO',
    'ISS': 'ISS',
    'PO': 'PO',
    'SSO': 'SSO',
    'ES-L1': 'ES-L1',
    'MEO': 'MEO',
    'GEO': 'GEO',
    'HEO': 'HEO',
    'SO': 'SO',
    'VLEO': 'VLEO'
}

# Apply mapping (keep original if not in mapping)
df_api['Orbit_clean'] = df_api['Orbit'].apply(lambda x: orbit_mapping.get(x, x) if pd.notna(x) else 'Unknown')

print("\nCleaned orbits:")
print(df_api['Orbit_clean'].value_counts())

# Cell 15: Feature Engineering - Create New Features
# Create additional useful features

# 1. Launch success rate by site (cumulative)
df_api = df_api.sort_values('Date')
df_api['Site_Success_Rate'] = df_api.groupby('LaunchSite')['Class'].cumsum() / df_api.groupby('LaunchSite').cumcount()

# 2. Days since first launch
first_launch_date = df_api['Date'].min()
df_api['Days_Since_First_Launch'] = (df_api['Date'] - first_launch_date).dt.days

# 3. Quarter of year
df_api['Quarter'] = df_api['Date'].dt.quarter

# 4. Payload weight category
def categorize_payload(mass):
    if pd.isna(mass):
        return 'Unknown'
    elif mass < 2000:
        return 'Light'
    elif mass < 5000:
        return 'Medium'
    elif mass < 10000:
        return 'Heavy'
    else:
        return 'Very Heavy'

df_api['Payload_Category'] = df_api['PayloadMass'].apply(categorize_payload)

print("\nNew features created:")
print("- Site_Success_Rate")
print("- Days_Since_First_Launch")
print("- Quarter")
print("- Payload_Category")

print("\nPayload categories:")
print(df_api['Payload_Category'].value_counts())

# Cell 16: Check Data Quality After Wrangling
print("\n" + "=" * 60)
print("DATA QUALITY AFTER WRANGLING")
print("=" * 60)

print("\nMissing values:")
print(df_api.isnull().sum().sum(), "total missing values")

print("\nDuplicates:")
print(df_api.duplicated().sum(), "duplicate rows")

print("\nData types:")
print(df_api.dtypes)

print(f"\nDataset shape: {df_api.shape}")

# Cell 17: Create Final Clean Dataset
# Select final columns for analysis
final_columns = [
    'FlightNumber',
    'Date',
    'Year',
    'Month',
    'Quarter',
    'LaunchSite',
    'PayloadMass',
    'Payload_Category',
    'Orbit',
    'Orbit_clean',
    'MissionOutcome',
    'LandingOutcome',
    'Class',
    'Days_Since_First_Launch'
]

# Keep only columns that exist
final_columns = [col for col in final_columns if col in df_api.columns]

df_clean = df_api[final_columns].copy()

print(f"\nFinal dataset shape: {df_clean.shape}")
print(f"Columns: {df_clean.columns.tolist()}")

display(df_clean.head(10))

# Cell 18: Statistical Summary
print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

# Numeric columns summary
print("\nNumeric features:")
print(df_clean.describe())

# Categorical columns summary
print("\nCategorical features:")
categorical_cols = df_clean.select_dtypes(include=['object']).columns
for col in categorical_cols:
    print(f"\n{col}:")
    print(df_clean[col].value_counts().head())

# Cell 19: Correlation Analysis
# Create correlation matrix for numeric features
numeric_df = df_clean.select_dtypes(include=[np.number])

plt.figure(figsize=(12, 8))
correlation_matrix = numeric_df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, fmt='.2f')
plt.title('Correlation Matrix - Numeric Features', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✓ Correlation matrix saved")

# Cell 20: Visualize Target Variable
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Get value counts
class_counts = df_clean['Class'].value_counts().sort_index()

# Bar chart
class_counts.plot(kind='bar', ax=axes[0], color=['#e74c3c', '#2ecc71'][:len(class_counts)])
axes[0].set_title('Landing Outcome Distribution', fontweight='bold', fontsize=12)
axes[0].set_xlabel('Class (0=Failure, 1=Success)')
axes[0].set_ylabel('Count')

# Dynamically set tick labels
tick_labels = ['Failure' if i == 0 else 'Success' for i in class_counts.index]
axes[0].set_xticks(range(len(class_counts)))
axes[0].set_xticklabels(tick_labels, rotation=0)

# Pie chart
class_counts.plot(kind='pie', ax=axes[1],
                  autopct='%1.1f%%',
                  colors=['#e74c3c', '#2ecc71'][:len(class_counts)],
                  labels=tick_labels)
axes[1].set_title('Landing Success Rate', fontweight='bold', fontsize=12)
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('target_variable_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

# Cell 21: Analyze Class Imbalance
class_counts = df_clean['Class'].value_counts()

# Safely get counts
class_0_count = class_counts.get(0, 0)
class_1_count = class_counts.get(1, 0)
total = len(df_clean)

print("\n" + "=" * 60)
print("CLASS IMBALANCE ANALYSIS")
print("=" * 60)

if total > 0:
    print(f"\nClass 0 (Failure): {class_0_count} ({class_0_count/total*100:.2f}%)")
    print(f"Class 1 (Success): {class_1_count} ({class_1_count/total*100:.2f}%)")
    
    if class_1_count > 0:
        imbalance_ratio = class_0_count / class_1_count
        print(f"\nImbalance ratio: {imbalance_ratio:.2f}")
        
        if imbalance_ratio > 1.5 or imbalance_ratio < 0.67:
            print("⚠️  Dataset is imbalanced. Consider using stratified sampling or class weights.")
        else:
            print("[OK] Dataset is relatively balanced.")
    else:
        print("\n⚠️ WARNING: No successful landings (Class 1) in dataset!")
else:
    print("\n⚠️ WARNING: Dataset is empty!")

print("=" * 60)

# Cell 22: Save Cleaned Data
# Save to CSV
df_clean.to_csv('data/spacex_cleaned.csv', index=False)
print("\n✓ Cleaned data saved to 'data/spacex_cleaned.csv'")

# DEBUG: Check Class column
print("=" * 60)
print("DEBUG: Class Column Analysis")
print("=" * 60)

print("\nClass value counts:")
print(df_clean['Class'].value_counts())

print("\nClass data type:")
print(df_clean['Class'].dtype)

print("\nUnique values in Class:")
print(df_clean['Class'].unique())

print("\nFirst 20 Class values:")
print(df_clean['Class'].head(20))

print("\nClass column null check:")
print(f"Null values: {df_clean['Class'].isnull().sum()}")

print("=" * 60)

# Cell 23: Create Data Wrangling Summary Report
class_counts = df_clean['Class'].value_counts()

# Safely get counts (default to 0 if class doesn't exist)
class_0_count = class_counts.get(0, 0)
class_1_count = class_counts.get(1, 0)
total = len(df_clean)

# Calculate percentages safely
class_0_pct = (class_0_count / total * 100) if total > 0 else 0
class_1_pct = (class_1_count / total * 100) if total > 0 else 0

summary_report = f"""
{'=' * 60}
DATA WRANGLING SUMMARY REPORT
{'=' * 60}

ORIGINAL DATA:
- Total records: {len(df_api)}
- Total columns: {len(df_api.columns)}

CLEANING STEPS PERFORMED:
1. [DONE] Handled missing values in PayloadMass (filled with median)
2. [DONE] Handled missing values in Orbit (filled with 'Unknown')
3. [DONE] Handled missing values in LandingOutcome
4. [DONE] Removed {duplicates} duplicate rows
5. [DONE] Converted data types (Date, numeric columns)
6. [DONE] Created target variable 'Class' (binary classification)
7. [DONE] Standardized launch site names
8. [DONE] Standardized orbit names
9. [DONE] Created new features (payload categories, quarters, etc.)

FINAL DATA:
- Total records: {len(df_clean)}
- Total columns: {len(df_clean.columns)}
- Missing values: {df_clean.isnull().sum().sum()}
- Duplicate rows: {df_clean.duplicated().sum()}

TARGET VARIABLE:
- Class 0 (Failure): {class_0_count} ({class_0_pct:.2f}%)
- Class 1 (Success): {class_1_count} ({class_1_pct:.2f}%)

DATE RANGE:
- From: {df_clean['Date'].min()}
- To: {df_clean['Date'].max()}

{'=' * 60}
"""

print(summary_report)

# Save report with UTF-8 encoding
with open('data_wrangling_report.txt', 'w', encoding='utf-8') as f:
    f.write(summary_report)

print("[DONE] Summary report saved to 'data_wrangling_report.txt'")

# Cell 24: Final Visualization Summary
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# 1. Launches over time
df_clean.groupby('Year').size().plot(kind='line', ax=axes[0, 0], marker='o', color='steelblue')
axes[0, 0].set_title('Launches Over Time', fontweight='bold')
axes[0, 0].set_xlabel('Year')
axes[0, 0].set_ylabel('Number of Launches')
axes[0, 0].grid(True, alpha=0.3)

# 2. Success rate over time
success_by_year = df_clean.groupby('Year')['Class'].mean() * 100
success_by_year.plot(kind='line', ax=axes[0, 1], marker='o', color='green')
axes[0, 1].set_title('Success Rate Over Time', fontweight='bold')
axes[0, 1].set_xlabel('Year')
axes[0, 1].set_ylabel('Success Rate (%)')
axes[0, 1].grid(True, alpha=0.3)

# 3. Launches by site
df_clean['LaunchSite'].value_counts().plot(kind='barh', ax=axes[0, 2], color='coral')
axes[0, 2].set_title('Launches by Site', fontweight='bold')
axes[0, 2].set_xlabel('Count')

# 4. Payload distribution by class
df_clean.boxplot(column='PayloadMass', by='Class', ax=axes[1, 0])
axes[1, 0].set_title('Payload Mass by Landing Outcome', fontweight='bold')
axes[1, 0].set_xlabel('Class (0=Failure, 1=Success)')
axes[1, 0].set_ylabel('Payload Mass (kg)')
plt.sca(axes[1, 0])
plt.xticks([1, 2], ['Failure', 'Success'])

# 5. Orbit distribution
df_clean['Orbit_clean'].value_counts().head(10).plot(kind='barh', ax=axes[1, 1], color='purple')
axes[1, 1].set_title('Top 10 Orbits', fontweight='bold')
axes[1, 1].set_xlabel('Count')

# 6. Payload categories
df_clean['Payload_Category'].value_counts().plot(kind='pie', ax=axes[1, 2], autopct='%1.1f%%')
axes[1, 2].set_title('Payload Categories', fontweight='bold')
axes[1, 2].set_ylabel('')

plt.suptitle('Data Wrangling Summary - SpaceX Falcon 9 Dataset', 
             fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.savefig('data_wrangling_summary.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n" + "=" * 60)
print("[DONE] DATA WRANGLING COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
